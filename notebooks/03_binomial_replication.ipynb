# Binomial replication

## European call replication

We build a portfolio replication to a call stock option payout. Assume the stock initially costs $ S_0 = 100 $ and at maturity of the option can be bought at either $ S_u = 120 $ or $ S_d = 80 $. The strike price of the call option is $ K = 100 $. From this we can write the call option payoff as

$$ C = \max{(S_{1} - K, 0)}~~,$$

where $ S_1 $ is the future price of the stock, that is either $ S_1 = S_u $ or $ S_1 = S_d $. 

We can build a portfolio of stocks and cash money that replicates these call payoffs. We can write

$$ \Delta S_u + 1.05 B = C_u = 20 \tag{1a}$$
$$ \Delta S_d + 1.05 B = C_d = 0  \tag{1b}$$

, where $ \Delta $ is the number of stocks and $ B $ is the money in cash we need to replicate the payoffs for the two possible scenarios and $ C_u,~C_d $ is the call option payoff for the up-state and low-state respectively. We have set the cash growth factor $ R = 1.05 $. We can solve this system of equations to get

$$ \Delta = \frac{1}{2},~~~ B = -\frac{800}{21} .$$

Using this we can evaluate what the cost of the call option will be assuming no-arbitrage rule. To do this we need to have the cost of financing our replication portfolio and the cost of just buying a call option be the same. Thus we can write

$$ C_0 = S_0 \left( \frac{1}{2} \right) + B = 100 \left( \frac{1}{2} \right) - \frac{800}{21} = \frac{250}{21} \approx 11.90 $$

, where $ C_0 $ is the cost of the call option. Now assume that there is an $ 80% $ chance the stock goes up and a $ 20% $ it goes low. We can write the expected payoff of the call option at maturity as

$$ \frac{4}{5}\,20 + \frac{1}{5}\,0 = 16 .$$

Now the replication price has not changed; indeed we never used any probability distribution to find $ C_0 $. Nonetheless, similar to what we did above, we can express that replication price as a discounted weighted average too. For that we are going to use a special weight $ q $, called the $ \textbf{risk-neutral probability} $. We choose it so that the stock's weighted average terminal value equals today;s price grown at the cash-account rate R:

$$ q S_u + (1 - q) S_d = 1.05 S_0 $$
$$ \implies q 120 + (1 - q) 80 = 1.05 (100) .$$

We solve this for to get

$$ q = \frac{5}{8} .$$

Under these pricing weights, the stock's expected return equals the cash-account return. Now we can use the same weights for the call payoffs that is

$$ q C_u + (1 - q) C_d = 1.05\;C_0 .$$

Solving for $ C_0 $ we see that this matches exactly the replication cost

$$ C_0 = \frac{(5/8) 20}{1.05} = \frac{250}{21} .$$

This is means that the risk-neutral weighted average is another way to express the price justified by replication. We can also see this by multiplying Eq $(1a)$ by $ q $ and Eq $(1b)$ by $ (1-q) $, adding them up and solving for the cost of replication gives

$$ \Delta S_0 + B = \frac{q C_u + (1 - q) C_d}{1.05} $$

as expected. A condition that is left to examine is we need $ 0 < q < 1 $ for positive probability, this implies

$$ S_d < 1.05 S_0 < S_u $$

, where the strict inequalities are required for the no-arbitrage condition.

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
src_path = project_root / "src"

if str(src_path) not in sys.path:
    sys.path.append(str(src_path))

import numpy as np
from options_lab.payoffs import call_payoff
from options_lab.payoffs import put_payoff

# Define variables 

S_0 = 100 # Initial stock price
S_u = 120 # Up-state stock price
S_d = 80  # Down-state stock price
K = 100   # Option strike price
R = 1.05  # Cash growth factor

assert S_d < R * S_0 < S_u  # No arbitrage-condition

C_u = call_payoff(S_u, K)
C_d = call_payoff(S_d, K)

Delta = (C_u - C_d)/ (S_u - S_d)
B = (C_d - Delta * S_d)/ R
C_0 = Delta * S_0 + B

q = (R * S_0 - S_d)/ (S_u - S_d)
C_0_rn = (q * C_u + (1 - q) * C_d)/ R

assert np.isclose(C_u, Delta * S_u + R * B)
assert np.isclose(C_d, Delta * S_d + R * B)
assert np.isclose(C_0, C_0_rn)


print(f"The call option pay off for the up-state is {C_u:.1f}")
print(f"The call option pay off for the down-state is {C_d:.1f}")
print(f"The amount of stocks needed for replication is {Delta:.1f}")
print(f"The cash-money needed for replication is {B:.3f}")
print(f"The initial portfolio cost is {C_0:.3f}")
print(f"Expected risk-neutral up probability is {q:.3f}")
```

## European put replication

A European put gives its holder the right to sell one share for the strike
price $K$ at maturity. Its payoff is

$$P_1 = \max(K-S_1,0).$$

With $K=100$, $S_u=120$ and $S_d=80$, the terminal payoffs are
$P_u=0$ and $P_d=20$, in euros.

To replicate these payoffs using shares and a cash account with growth
factor $R=1.05$, we require

$$120\Delta+1.05B=0,$$
$$80\Delta+1.05B=20.$$

Solving gives

$$\Delta=-\frac12,\qquad B=\frac{60}{1.05}\approx57.1429.$$

We therefore short half a share and deposit enough money today to receive
€60 at maturity. After buying back the half-share, the portfolio leaves
€0 in the up state or €20 in the down state.

The initial cost determines the put's no-arbitrage price:

$$P_0=\Delta S_0+B=-50+\frac{60}{1.05}
=\frac{50}{7}\approx 7.1429\text{ euros}.$$

Using the same risk-neutral up probability $q=5/8$ also gives

$$P_0=\frac{qP_u+(1-q)P_d}{R}
=\frac{(3/8)\,20}{1.05}.$$

The call and put prices satisfy put–call parity:

$$C_0-P_0=S_0-\frac{K}{R}.$$

```python
def price_one_period_call(S_0, S_u, S_d, K, R):

    assert S_d < R * S_0 < S_u

    C_u = call_payoff(S_u, K)
    C_d = call_payoff(S_d, K)

    Delta = (C_u - C_d)/ (S_u - S_d)
    B = (C_d - Delta * S_d)/ R
    C_0 = Delta * S_0 + B

    q = (R * S_0 - S_d)/ (S_u - S_d)
    C_0_rn = (q * C_u + (1 - q) * C_d)/ R

    assert np.isclose(C_u, Delta * S_u + R * B)
    assert np.isclose(C_d, Delta * S_d + R * B)
    assert np.isclose(C_0, C_0_rn)

    return Delta, B, q, C_0, C_0_rn

# Strike price higher than both terminal stock prices K = 130
assert price_one_period_call(100, 120, 80, 130, 1.05)[0] == price_one_period_call(100, 120, 80, 130, 1.05)[1] == 0

def price_one_period_put(S_0, S_u, S_d, K, R):

    assert S_d < R * S_0 < S_u

    P_u = put_payoff(S_u, K)
    P_d = put_payoff(S_d, K)

    Delta = (P_u - P_d)/ (S_u - S_d)
    B = (P_d - Delta * S_d)/ R
    P_0 = Delta * S_0 + B

    q = (R * S_0 - S_d)/ (S_u - S_d)
    P_0_rn = (q * P_u + (1 - q) * P_d)/ R

    assert np.isclose(P_u, Delta * S_u + R * B)
    assert np.isclose(P_d, Delta * S_d + R * B)
    assert np.isclose(P_0, P_0_rn)

    return Delta, B, q, P_0, P_0_rn

Call_0 = price_one_period_call(100, 120, 80, 100, 1.05)[3]
Put_0 = price_one_period_put(100, 120, 80, 100, 1.05)[3]

assert np.isclose(Call_0 - Put_0, S_0 - K/R)
```

```python
# Writing the two functions above as one. Now instead of using the call and put functions we use the payoffs as inputs directly.

def replicate_one_period(S_0, S_u, S_d, payoff_u, payoff_d, R):
    
    ' Asks how much money is needed now to build a portfolio that will deliver € (payoff_u) or € (payoff_d) one period later that'

    assert S_d < R * S_0 < S_u
    
    Delta = (payoff_u - payoff_d)/ (S_u - S_d)
    B = (payoff_d - Delta * S_d)/ R
    price_0 = Delta * S_0 + B
    q = (R * S_0 - S_d)/ (S_u - S_d)
    price_0_rn = (q * payoff_u + (1 - q) * payoff_d)/ R

    assert np.isclose(payoff_u, Delta * S_u + R * B)
    assert np.isclose(payoff_d, Delta * S_d + R * B)
    assert np.isclose(price_0_rn, price_0)

    return Delta, B, q, price_0, price_0_rn

print(replicate_one_period(100, 120, 80, 20, 0, 1.05))
print(replicate_one_period(100, 120, 80, 0, 20, 1.05))
```

```python
# Calculating cost of replication portfolio with two time steps.

S_0 = 100
K = 100
R = 1.05
u = 1.2
d = 0.8

# At t = 0

# At t = 1
# up:   S_1 = 120 
# down: S_1 = 80

# At t = 2 terminal prices are:
# up-up:                     S_2 = 144
# up-down (same as down-up): S_2 = 96
# down-down:                 S_2 = 64

# Holdings and option values at t = 1, replicating payoffs at t = 2
up_node = replicate_one_period(S_0 * u, S_0 * u * u, S_0 * u * d, 
            call_payoff(S_0 * u * u, K), 
            call_payoff(S_0 * u * d, K), 1.05)

down_node = replicate_one_period(S_0 * d, S_0 * u * d, S_0 * d * d, 
            call_payoff(S_0 * d * u, K), 
            call_payoff(S_0 * d * d, K), 1.05)

# Option value at t = 1 after an up move
value_up = up_node[3]

# Option value at t = 1 after a down move
value_down = down_node[3]

# Initial holdings and price at t = 0, replicating values at t = 1
call_results = replicate_one_period(S_0, S_0 * u, S_0 * d, 
            value_up, value_down, 
            1.05)


# Number of shared held after each move
Delta_u = up_node[0]
Delta_d = down_node[0]

# Number of shares needed to finance replicating portfolio at t = 0
Delta_0 =  call_results[0]
q = call_results[2]

# Cost of financing replicating portfolio at t = 0
C_0 =  call_results[3]


C_0_formula = (44 * q**2)/ R**2
assert np.isclose(C_0, C_0_formula)

print(Delta_u)
print(Delta_d)
print(Delta_0)
print(f"Financing cost for replication portfolio is {C_0:.3f}")
```
