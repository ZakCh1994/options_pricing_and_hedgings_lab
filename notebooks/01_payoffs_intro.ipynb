# Basic Payoff Diagrams

This notebook visualises basic payoff profiles of basic derivative contracts:

- long call option
- long put option
- short call option
- short put option
- long forward
- short forward

The payoff functions are imported from the reusable source code in 'srs/options_lab/payoff.py'

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
src_path = project_root / "src"

if str(src_path) not in sys.path:
    sys.path.append(str(src_path))

import numpy as np
import matplotlib.pyplot as plt

from options_lab.payoffs import(
    call_payoff,
    put_payoff,
    long_forward_payoff,
    short_forward_payoff,
    short_call_payoff,
    short_put_payoff,
)
```

```python
strike = 100
delivery_price = 100

spots = np.linspace(50, 150, 101)
```

```python
long_call = call_payoff(spots, strike)
long_put = put_payoff(spots, strike)

short_call = short_call_payoff(spots, strike)
short_put = short_put_payoff(spots, strike)

long_forward = long_forward_payoff(spots, delivery_price)
short_forward = short_forward_payoff(spots, delivery_price)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(spots, long_call, label="Long call")
plt.plot(spots, short_call, label="Short call")

plt.axhline(0, linewidth=1)
plt.axvline(strike, linestyle="--", linewidth=1, label="Strike")

plt.xlabel("Terminal spot price")
plt.ylabel("Payoff")
plt.title("Call Option Payoffs")
plt.legend()
plt.grid(True)

plt.show
```

```python
plt.figure(figsize=(8, 5))

plt.plot(spots, long_put, label="Long put")
plt.plot(spots, short_put, label="Short put")

plt.axhline(0, linewidth=1)
plt.axvline(strike, linestyle="--", linewidth=1, label="Strike")

plt.xlabel("Terminal spot price")
plt.ylabel("Payoff")
plt.title("Put Option Payoffs")
plt.legend()
plt.grid(True)

plt.show

```

```python
plt.figure(figsize=(8, 5))

plt.plot(spots, long_forward, label="Long forward")
plt.plot(spots, short_forward, label="Short forward")

plt.axhline(0, linewidth=1)
plt.axvline(delivery_price, linestyle="--", linewidth=1, label="Delivery price")

plt.xlabel("Terminal spot price")
plt.ylabel("Payoff")
plt.title("Forward Payoffs")
plt.legend()
plt.grid(True)

plt.show()
```
